```python
from langgraph.graph import StateGraph, START, END
```

```python
from typing import TypedDict
```

```python
class State(TypedDict):
    message: str
```

```python
def hello_node(state: State):
    print("Hello Node")

    return state
```

```python
def bye_node(state: State):
    print("Bye world")

    return state
```

```python
graph = StateGraph(State)

graph.add_node("hello", hello_node)
graph.add_node("bye", bye_node)

graph.add_edge(START, "hello")
graph.add_edge("hello", "bye")
graph.add_edge("bye", END)
```

```python
graph = graph.compile()
```

```python
graph.invoke(
    {
        "message": "Hi"
    }
)
```

```python
graph.invoke(
    {
        "message": "How are you?"
    }
)
```
